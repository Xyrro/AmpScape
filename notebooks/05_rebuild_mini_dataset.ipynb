# Rebuild the mini dataset locally (needs Julia + Circuitscape)

The pipeline is `plan → prepare → solve → finalize` (`scripts/generate.py`). This rebuilds a 5-sample smoke build; the full mini is the first 3 S shards of the v1.0 stream (`scripts/plan_v1.py --tier S --n 600`).

```python
import subprocess, sys
run = lambda *a: subprocess.run([sys.executable, 'scripts/generate.py', *a], check=True)
run('plan', '--dataset', 'smoke', '--out', 'data/builds/nb_smoke', '--n-synthetic', '5', '--n-real', '0', '--tier', 'S', '--shard-size', '5')
run('prepare', '--build', 'data/builds/nb_smoke')
run('solve', '--build', 'data/builds/nb_smoke', '--shard', '0')      # local Julia solve (minutes)
run('finalize', '--build', 'data/builds/nb_smoke', '--quicklooks')
```
