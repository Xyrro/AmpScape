# AmpScape quickstart

Load the **mini** subset (3 S shards, all tasks, ≈ 0.4 GB) straight from the Hub, look at one sample, and read the index. Requires `uv sync` (no Julia).

```python
from ampscape.data import load_from_hub
ds = load_from_hub('T1', 'S', split='train', subset='mini')
print(len(ds), 'items;', ds.index.columns.tolist()[:12])
```

```python
x = ds[0]
for k, v in x.items():
    print(k, getattr(v, 'shape', type(v).__name__))
print(x['meta']['generator'], x['meta'].get('contrast'), x['meta']['tier'])
```

```python
import numpy as np, matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].imshow(np.log10(x['resistance'][0]), cmap='magma'); ax[0].set_title('log10 R')
ax[1].imshow(x['focal'][0] > 0, cmap='gray'); ax[1].set_title('focal nodes')
ax[2].imshow(np.log10(x['cum_current'][0] + 1e-6 * x['cum_current'].max()), cmap='viridis'); ax[2].set_title('log10 cumulative current')
for a in ax: a.axis('off')
```

Splits and OOD flags are columns of the index: `split`, `test_ood_region`, `test_ood_table`, `test_ood_contrast`, `test_ood_scale`, `test_ood_published`, `test_ood_scale_strict`, and the subset memberships `subset_mini / core / full`.
