# Evaluate predictions and compare models

Any predictions directory in the documented format (`docs/evaluation.md`) can be scored with `scripts/evaluate.py`; here we compare a trained model with the non-learned coarsen-×4 baseline on the same split.

```python
import subprocess, sys
root = 'PATH_TO_LOCAL_ROOT'
subprocess.run([sys.executable, 'scripts/evaluate.py', '--predictions', 'runs/nb/unet_T1/predictions/S_S_test_id', '--split', 'test_id', '--root', root], check=True)
```

```python
import json
r = json.load(open('runs/nb/unet_T1/predictions/S_S_test_id/eval_test_id/results.json'))
{k: round(v['mean'], 4) for k, v in r['per_task']['T1'].items() if isinstance(v, dict) and 'mean' in v}
```

Metric definitions: pixel (log10-ε MAE, relative L2), domain (top-q % IoU, pinch-point recall, corridor Dice, Spearman), Reff (T2), physics (Kirchhoff residual with a voltage prediction, non-negativity, focal current), speed-up, and the solver-acceleration track (`--acceleration`, warm-started AMG-PCG). Thresholds are conventions.
