# Train the official U-Net baseline (T1, tier S)

Uses `scripts/train.py` with the frozen official configuration; on the mini subset a few epochs take minutes on one GPU (works on CPU, slowly). Inputs: standardised log-resistance, NoData mask, focal mask; target: log10(C + ε·max C); masked MSE.

```python
import subprocess, sys
root = 'PATH_TO_LOCAL_ROOT'   # e.g. the snapshot_download directory printed by load_from_hub, or data/builds/mini
subprocess.run([sys.executable, 'scripts/train.py', '--model', 'unet', '--task', 'T1', '--tier', 'S', '--root', root, '--out', 'runs/nb/unet_T1', '--epochs', '10', '--batch', '16', '--eval-splits', 'test_id', '--published-root', 'none'], check=True)
```
