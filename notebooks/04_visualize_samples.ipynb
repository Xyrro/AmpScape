# Visualise samples across tasks and tiers

```python
from ampscape.data import load_from_hub
import numpy as np, matplotlib.pyplot as plt
for task, key in [('T1', 'cum_current'), ('T3', 'current'), ('T4', 'cum_current')]:
    ds = load_from_hub(task, 'S', split='test_id', subset='mini')
    x = ds[0]
    fig, ax = plt.subplots(1, 2, figsize=(8, 4))
    ax[0].imshow(np.log10(x['resistance'][0]), cmap='magma'); ax[0].set_title(f'{task} log10 R')
    ax[1].imshow(np.log10(x[key][0] + 1e-6 * np.nanmax(x[key])), cmap='viridis'); ax[1].set_title(key)
    for a in ax: a.axis('off')
```

Every sample's `meta` records the generator or tile provenance, the resistance table, the solver preset, versions, QC flags and the pipeline tag; `solver_stats` on each configuration group records timings, residuals and (test/OOD) the CG baseline.
